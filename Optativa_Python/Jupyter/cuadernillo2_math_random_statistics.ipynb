## 2. El módulo math
| Función | Qué hace | Ejemplo |
|---|---|---|
| `math.sqrt(x)` | Raíz cuadrada | `sqrt(16)` → `4.0` |
| `math.pow(x, y)` | Potencia (float) | `pow(2, 3)` → `8.0` |
| `math.ceil(x)` | Redondea hacia arriba | `ceil(4.1)` → `5` |
| `math.floor(x)` | Redondea hacia abajo | `floor(4.9)` → `4` |
| `math.fabs(x)` | Valor absoluto (float) | `fabs(-3)` → `3.0` |
| `math.factorial(n)` | Factorial | `factorial(5)` → `120` |
| `math.gcd(a, b)` | Máximo común divisor | `gcd(48, 18)` → `6` |
| `math.log(x, base)` | Logaritmo | `log(8, 2)` → `3.0` |
| `math.sin(x)`, `math.cos(x)` | Trigonometría (radianes) | `cos(0)` → `1.0` |
| `math.radians(g)` | Grados a radianes | `radians(180)` |
| `math.hypot(a, b)` | Hipotenusa | `hypot(3, 4)` → `5.0` |

````python
import math

print(math.pi, math.e)
radio = 5
area = math.pi * radio ** 2
print(round(area, 2))   # 78.54
print(math.sqrt(16), math.ceil(4.1), math.floor(4.9), math.gcd(48, 18), math.hypot(3, 4))
````

**Ojo:** `round()` redondea al valor más cercano; `math.ceil` y `math.floor` siempre van hacia arriba o hacia abajo.

## 3. El módulo random
| Función | Qué hace |
|---|---|
| `random.random()` | Decimal entre 0 y 1 (sin incluir 1) |
| `random.randint(a, b)` | Entero entre a y b, **ambos incluidos** |
| `random.uniform(a, b)` | Decimal entre a y b |
| `random.choice(lista)` | Un elemento al azar |
| `random.choices(lista, k=3)` | k elementos **con** repetición |
| `random.sample(lista, 3)` | k elementos **sin** repetición |
| `random.shuffle(lista)` | Desordena la lista (modifica la original) |
| `random.seed(n)` | Hace la secuencia reproducible |

````python
import random

print(random.randint(1, 6))                          # un dado
print(random.choice(["piedra", "papel", "tijera"]))

baraja = [1, 2, 3, 4, 5]
random.shuffle(baraja)
print(baraja)

random.seed(42)
print(random.randint(1, 100))   # siempre igual con esta semilla
````

**Importante:** `random` no es seguro para contraseñas ni tokens; para eso existe el módulo `secrets`.

## 4. El módulo statistics
| Función | Descripción |
|---|---|
| `mean(datos)` | Media |
| `median(datos)` | Mediana (con cantidad par promedia los dos centrales) |
| `mode(datos)` | Moda; `multimode` devuelve todas si hay empate |
| `stdev` / `pstdev` | Desviación típica de muestra / población |
| `variance` / `pvariance` | Varianza de muestra / población |

**Aviso:** con una lista vacía dan `StatisticsError`. Comprueba `if datos:` antes.

````python
import statistics

notas = [4, 7, 7, 8, 10]
print(statistics.mean(notas))      # 7.2
print(statistics.median(notas))    # 7
print(statistics.mode(notas))      # 7
print(statistics.stdev(notas))
print(statistics.variance(notas))
````

## 5. Mini proyecto: simulador de dados

````python
tiradas = [random.randint(1, 6) for _ in range(100)]
print("Media:", round(statistics.mean(tiradas), 2))
print("Moda:", statistics.mode(tiradas))
print("Veces que salió 6:", tiradas.count(6))
````
