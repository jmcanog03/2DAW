# Cuadernillo 3: NumPy

**Objetivos:** crear y manipular arrays, operaciones vectorizadas, indexado, slicing, filtrado con máscaras y agregaciones por ejes.

## 1. ¿Qué es NumPy?
Librería base de cálculo numérico. Su objeto principal es el **array** (`ndarray`): elementos **del mismo tipo**, almacenados de forma compacta y con operaciones **vectorizadas** (sin bucles). Es la base de pandas y matplotlib.

````python
# !pip install numpy
import numpy as np
````

## 2. Lista frente a array

````python
lista = [1, 2, 3]
array = np.array([1, 2, 3])

print(lista * 2)    # [1, 2, 3, 1, 2, 3]  (repite la lista)
print(array * 2)    # [2 4 6]             (multiplica cada elemento)
````

## 3. Crear arrays

````python
print(np.array([[1, 2, 3], [4, 5, 6]]))
print(np.zeros(5))
print(np.ones((2, 3)))
print(np.full((2, 2), 7))
print(np.arange(0, 10, 2))      # [0 2 4 6 8]  controla el PASO
print(np.linspace(0, 1, 5))     # 5 valores, INCLUYE el final
print(np.random.randint(1, 10, 5))
````

## 4. Atributos y forma

````python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.shape, a.ndim, a.size, a.dtype)
print(np.arange(6).reshape(2, 3))
````

## 5. Indexado y slicing: `a[filas, columnas]`
Un slice es una **vista**, no una copia. Para copiar usa `a.copy()`.

````python
a = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])
print(a[0, 0])      # 10
print(a[1, 2])      # 60
print(a[0])         # primera fila
print(a[:, 1])      # segunda columna
print(a[0:2, 1:])   # subtabla
````

## 6. Operaciones vectorizadas

````python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])
print(x + y, x * y, x ** 2, x > 2)
print(np.sqrt(x), np.exp(x), np.log(x), np.sin(x))
````

## 7. Agregaciones y ejes
`axis=0` colapsa las filas (un resultado por **columna**); `axis=1` colapsa las columnas (un resultado por **fila**).

````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m.sum(), m.mean(), m.max(), m.min())
print(m.sum(axis=0))   # [5 7 9]
print(m.sum(axis=1))   # [ 6 15]
````

Otras funciones útiles: `np.median`, `np.std`, `np.argmax`, `np.argmin`, `np.sort`, `np.cumsum`.

## 8. Filtrado con máscaras booleanas
Combina condiciones con `&` (y), `|` (o) y paréntesis. **No** uses `and`/`or` (da `ValueError`).

````python
edades = np.array([15, 22, 17, 40, 18, 9])
mascara = edades >= 18
print(mascara)
print(edades[mascara])                         # [22 40 18]
print(np.sum(edades >= 18))                    # 3
print(edades[(edades >= 18) & (edades < 30)])  # [22 18]
````

## 9. Matrices

````python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(A * B)   # elemento a elemento
print(A @ B)   # producto matricial
print(A.T)     # traspuesta
````

## 10. `np.where`

````python
notas = np.array([3, 8, 5, 9, 4])
print(np.where(notas >= 5, "Aprobado", "Suspenso"))
````
