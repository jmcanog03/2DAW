**Objetivos:** usar `colorama` para dar color y estilo al texto de la consola, resetear estilos y aplicarlo a mensajes de error, éxito y aviso.

## 1. ¿Qué es colorama?
Permite cambiar el **color del texto, el fondo y el estilo** de lo que se imprime en consola, y funciona también en Windows.

````python
# Ejecuta esta celda solo si no tienes colorama instalado
# !pip install colorama
````

## 2. Las tres clases principales
| Clase | Para qué sirve | Ejemplos |
|---|---|---|
| `Fore` | Color del texto | `RED`, `GREEN`, `YELLOW`, `BLUE`, `MAGENTA`, `CYAN`, `WHITE`, `BLACK` |
| `Back` | Color de fondo | `Back.RED`, `Back.GREEN`... |
| `Style` | Estilo | `BRIGHT`, `DIM`, `NORMAL`, `RESET_ALL` |

> **Nota para Jupyter:** en un script normal se suele usar `init(autoreset=True)`. En un cuaderno es más seguro **no** llamar a `init()` y cerrar cada texto con `Style.RESET_ALL`. Si los colores no se ven en tu entorno, ejecútalo en una terminal o en VS Code.

````python
from colorama import Fore, Back, Style

print(Fore.RED + "Esto es un error" + Style.RESET_ALL)
print(Fore.GREEN + "Esto es correcto" + Style.RESET_ALL)
print("Texto normal")
````

## 3. Resetear colores
El color **se queda activo** hasta que se cambia o se resetea. Dos formas:
- Manual: `print(Fore.RED + "Error" + Style.RESET_ALL)`
- Automática (en scripts): `init(autoreset=True)`

## 4. Combinar color, fondo y estilo

````python
print(Fore.WHITE + Back.RED + "PELIGRO" + Style.RESET_ALL)
print(Style.BRIGHT + Fore.CYAN + "Texto brillante" + Style.RESET_ALL)
print(Style.DIM + "Texto tenue" + Style.RESET_ALL)
````

## 5. Ejemplo realista: funciones de mensajes

````python
def ok(mensaje):
    print(Fore.GREEN + "[OK] " + mensaje + Style.RESET_ALL)

def aviso(mensaje):
    print(Fore.YELLOW + "[AVISO] " + mensaje + Style.RESET_ALL)

def error(mensaje):
    print(Fore.RED + Style.BRIGHT + "[ERROR] " + mensaje + Style.RESET_ALL)

ok("Usuario creado")
aviso("La contraseña es corta")
error("No se pudo conectar con la base de datos")
````

**Buenas prácticas:** no abuses de los colores y dales significado coherente (rojo = error, verde = éxito, amarillo = aviso). Si rediriges la salida a un fichero se guardan los códigos de escape.

````python
# Tu solución C1

color = input("Introduce un color solo puede ser rojo, amarillo o verde")

match color:
    case "rojo":
        print(Fore.RED + color + Style.RESET_ALL)

    case "amarillo":
        print(Fore.YELLOW + color + Style.RESET_ALL)

    case "verde":
        print(Fore.GREEN + color + Style.RESET_ALL)
        
    case _:
        print(Fore.RED + Style.BRIGHT +color + Style.RESET_ALL)

````

````python
# C1
color = input("Color (rojo/amarillo/verde): ").lower()
if color == "rojo":
    print(Fore.RED + "rojo" + Style.RESET_ALL)
elif color == "amarillo":
    print(Fore.YELLOW + "amarillo" + Style.RESET_ALL)
elif color == "verde":
    print(Fore.GREEN + "verde" + Style.RESET_ALL)
else:
    print(Fore.RED + Style.BRIGHT + "Color no válido" + Style.RESET_ALL)
````

````python
# C3
COLORES = {"INFO": Fore.BLUE, "WARN": Fore.YELLOW, "ERROR": Fore.RED}

def log(nivel, mensaje):
    color = COLORES.get(nivel, Fore.WHITE)
    print(color + f"[{nivel}] {mensaje}" + Style.RESET_ALL)

log("INFO", "Servidor iniciado")
log("ERROR", "Fallo de conexión")
````
